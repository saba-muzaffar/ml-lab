Ploting with matplotlib

```python
import matplotlib.pyplot as plt
X = [1,2,3,4]
Y = [10,20,30,40]
plt.plot(X, Y, marker='o')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Simple Plot')
plt.show()
```
